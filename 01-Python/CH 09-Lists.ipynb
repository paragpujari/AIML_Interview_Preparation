##### Lists

1.  A List is a fundamental data structure in python that has elements enclosed with square brackets [].

2.  It may or may not have duplicate elements.

3.  It is ordered, mutuable(elements in the lists can be changed).

4.  It is indexed and can be accesed by means of an index position starting from 0 till (n-1).

5.  It can have the elements of same or different types. 

##### Creation of Lists

```python
##### empty lists
empty_list = []
print(empty_list)
print(type(empty_list))
```

```python
##### lists with same type of elements
list1 = [1,2,3,4,5]
print(list1)
print(type(list1))
```

```python
##### lists with different types of elements
list1 = [11,23.56,'Parag',True,(3+4j),[22,33],(45,56),{'name':'Anish','age':31}]

print(list1)

print(type(list1))
```

```python
##### Lists can have duplicates
lists = [10,10,20,20,20,30,30,40,40,50]

print(lists)

print(type(lists))
```

```python
##### Indexing
lists = [10,10,20,20,20,30,30,40,40,50]

print(lists[0])
print(lists[-1])
```

##### OBSERVATIONS:

1.  The lists start with an index position 0 and ends till (n-1)

```python
numbers = [10, 20, 30, 40, 50]

print('Positive Indexing')
print(numbers[0])
print(numbers[1])
print(numbers[2])
print(numbers[3])
print(numbers[4])

print('----------------------------------------------')
print('Negative Indexing')
print(numbers[-1])
print(numbers[-2])
print(numbers[-3])
print(numbers[-4])
print(numbers[-5])
```

##### Slicing

It refers to the technique of extracting the portion of list from the entire list.

```python
numbers = [10, 20, 30, 40, 50]

##### Get the elements at the even position
print(numbers[0:5:2])

##### Get the elements at the odd position
print(numbers[1:5:2])
```

##### Modifying the elements in the lists

1.  Lists are mutuable. The elements in the lists can be changed at any index position.

```python
numbers = [10, 20, 35, 40, 50]

print('---------------------------Original List----------------------------------')
print(numbers)

##### Change the element 35 -----> 30 at index 2
numbers[2] = 30

print('--------------------------Modified List-----------------------------------')
print(numbers)
```

##### OBSERVATIONS:

1.  As the list is mutuable, so the element 35 at index position 2 is changed to 30.

##### Operations in Lists

(a.)    append  ----->  It adds an element at the end of the list.

```python
num = [10,20,30,40]
print('---------------------------Original List----------------------------------')
print(num)

##### Add an element 50 at the end
num.append(50)

print('---------------------------Modified List----------------------------------')
print(num)
```

(b.)    insert  -----> It adds an element at the specific index position

```python
num = [10,30,40]
print('---------------------------Original List----------------------------------')
print(num)

print('---------------------------Modified List-----------------------------------')
##### Add element 20 at index position 1
num.insert(1,20)
print(num)
```

(c.) extend -----> It adds one list at the end of another list

```python
num1 = [10,20]

num2 = [30,40]

##### Adding num2 at the end of num1
num1.extend(num2)

print(num1)
```

##### Removal of element from the list

(a.)    remove ----->   It deletes an element from the list based on the value

```python
num = [10,20,30,40,50]

##### Remove element 50 from the list
num.remove(50)

print(num)
```

(b.)    pop -----> It deletes an element from the list based on the index position

```python
num = [10,20,30,40,50]

##### Remove the element at index position 1

num.pop(1)

print(num)
```

(c.) clear -----> It deletes all the element from the list

```python
num = [10,20,30,40,50]

num.clear()

print(num)
```

##### List Searching

It refers to the technique of finding out the particular element from the list

```python
num  = [1,2,3,4,5]

print(1 in num)
print(10 not in num)
```

```python
##### index -----> It finds out the index position for the particular element in the list
num = [1,2,3,4,5]
print(num.index(1))
```

```python
###### count -----> It counts the total number of occurances for a particular element in the list
num = [1,2,3,4,5]
print(num.count(1))
```

```python
##### List length -----> It finds out the total length of the list
num = [10,20,30,40,50]
l = len(num)
print(f'The total length of the list is {l}')
```

```python
##### Sorting -----> It is used to arrange the elements either in ascending or in descending order
numbers = [2,1,4,3,5]
print(numbers)
###### Arrange the elements in ascending order
numbers.sort()
print(numbers)

##### Arrange the elements in descending order
numbers.sort(reverse=True)
print(numbers)
```

```python
##### Reverse -----> It is used to reverse the elements of the list
numbers = [1,2,3,4,5]
print(numbers)
print('---------------------------------------')
numbers.reverse()
print(numbers)
```

```python
##### copy -----> It is used to copy the contents from one list to another
list1 = [1,2,3,4,5]

print(list1)

list2 = list1.copy()

print(list2)
```

```python
##### List concatenation -----> It is used to join two lists
list1 = [1,2,3,4,5]
list2 = [6,7,8,9,10]

print(list1 + list2)
```

```python
##### Replication -----> It is used to repeat the list for 'n' number of times
lists = [1,2,3,4,5]
print(lists*2)
```

```python
##### List Traversal -----> It is used to iterate through all the elements of the list
lists = [1,2,3,4,5]
for i in lists:
    print(i)
```

11. Write a Python program to separate positive and negative numbers into two different lists

```python
data = [10, -5, 20, -8, 30, -2, 0]

positive=[]
negative=[]

for x in data:
    if(x>0):
        positive.append(x)
    elif(x==0):
        continue
    else:
        negative.append(x)

print(positive)
print(negative)
```

13. Write a Python program to find the missing number from a list containing numbers from 1 to N

```python
data = [1, 2, 3, 5, 6, 7, 8]

actual=0
for i in data:
    actual += i
print(f'Actual data is {actual}')


l=len(data)
print(l)
expected=(l*(l+1))/2
print(f'Expected value is {expected}')

##### Calculate the missing value
missing = actual - expected
print(f'Missing value is {actual-expected}')
```

14. Write a Python program to identify the three most frequently occurring elements in a list

```python
data = [10, 20, 10, 30, 20, 10, 40, 20, 30, 10]

freq={}

for x in data:
    if(x in freq):
        freq[x] = freq[x]+1
    else:
        freq[x] = 1
print(freq)

##### Sort the items in the dictionary
ans = sorted(freq.items(), key= lambda x: x[1], reverse=True)
print(ans[:3])
```

15. Write a Python program to convert a nested list into a single-level list

```python
data = [1, [2, 3], [4, [5, 6]], 7]

ans=[]
for x in data:
    if(isinstance(x,list)):
        ans.extend(x)
    else:
        ans.append(x)
#print(ans)

ans1=[]
for x in ans:
    if(isinstance(x,list)):
        ans1.extend(x)
    else:
        ans1.append(x)
print(ans1)
```

16. Write a Python program to identify all pairs of elements whose sum equals a given target

```python
data = [2, 4, 3, 5, 7, 8, 1]
target = 9

for x in range(len(data)):
    for y in range(x+1,len(data)):
        if(data[x]+data[y]==target):
            print(data[x],data[y])
```

19. Write a Python program to find the longest sequence of consecutive integers in a list

```python
data = [100, 4, 200, 1, 3, 2, 5, 10]

data.sort()

print(data)

l=0

for x in range(len(data)-1):
    if(data[x]+1==data[x+1]):
        print(data[x])
        l +=1

print(f'Length of the longest consecutive sequence of integers in the list is {l}')
```

20. Write a Python program to identify the three most frequently occurring elements in a list

```python
from collections import Counter

data = [10, 20, 10, 30, 20, 10, 40, 20, 30, 10]

##### To get the frequency of all the elements in the list
frequencies = Counter(data)

print(frequencies)

##### To get the three most frequently occuring elements in the list
print(frequencies.most_common(3))
```

##### Analytical Scenario Based Questions

1.  Answer the following interview questions:

(a.)     How would you identify the highest and lowest employee salaries?

(b.)     How would you calculate the average salary?

(c.)     How would you identify the second-highest distinct salary?

(d.)     How would you find the most frequently occurring salary?

(e.)     How would you identify employees earning more than ₹60,000?

```python
salaries = [
    30000, 45000, 50000, 30000, 60000,
    75000, 45000, 90000, 30000, 100000,
    50000, 45000, 80000, 120000, 60000
]

##### highest salary

highest_salary = max(salaries)

print(f'Highest salary of the employee is {highest_salary}')


##### lowest salary

lowest_salary = min(salaries)

print(f'Lowest salary of the employee is {lowest_salary}')


##### average salary

average_salary = sum(salaries)/len(salaries)

print(f'Average salary of the employee is {average_salary}')

##### second-highest distinct salary

ans = [x for x in salaries if (x != highest_salary)]

print(ans)

print(f'Second highest salaery of an employee is {max(ans)}')


##### most frequently occurring salary

from collections import Counter

frequencies = Counter(salaries)

print(frequencies.most_common(1))

##### employees earning more than ₹60,000
ans = [x for x in salaries if(x>60000)]

print(ans)
```
