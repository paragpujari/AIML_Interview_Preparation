##### Operators

1.  An operator is a symbol that is used to perform various operations like mathematical/comparison/logical etc in the program.


##### Types of Operators

##### Main Inbuilt Data Types

| Category    | Operators                 |   Purpose                       |
| ------------| --------------------------|---------------------------------|
| Arithmetic  | + - * / // % **           |   Mathematical calculations     |
| Comparison  | == != > < >= <=           |   Compare Values                |
| Assignment  | = += -= *= /= //= %= **=  |   Assign or update the values   |
| Logical     | and or not                |   Combine conditions            |
| Bitwise     | & \| ^ ~ << >>            |   Work at bit level             |
| Membership  | in, not in                |   Check membership              |
| Identity    | is, is not                |   Check object identity         |


##### Types of Operators

(a.)    Arithmetic Operators ------> These are the operators that is used to perform mathematical operations on the data.

```python
a=10
b=3

print(f"Addition is {a+b}")
print(f"Substraction is {a-b}")
print(f"Multiplication is {a*b}")
print(f"Division is {a/b}")
print(f"Modulus is {a%b}")
print(f"Floor value is {a//b}")
print(f"Exponent is {a**b}")
```

(b.) Comparision Operators --------> These are the operators that is used to compare between two values

```python
a=10
b=3

print(a>b)  ##### True
print(a<b)  ##### False
print(a>=b) ##### True
print(a<=b) ##### False
print(a==b) ##### False
print(a!=b) ##### True
```

(c.) Assignment Operators ---------> These are the operators that is used to assign or update the values

```python
##### perform assignment operators
a=10
a+=3    ##### ----->    a+=3 => a=a+3
print(a)

b=10
b-=3    ##### ----->    b-=3 => b=b-3
print(b)

c=10
c*=3    ##### ----->    c*=3 => c=c*3
print(c)

d=10
d/=3    ##### ----->   d/=3 => d=d/3
print(d)

e=10
e//=3   ##### ----->    e//=3 => e=e//3
print(e)

f=10
f%=3    ##### ----->    f%=3 => f=f%3
print(f)

g=10
g**=3    ##### ----->   g**=3 => g=g**3
print(g)
```

(d.) Logical Operators -------> It is the operator that is used to combine the values. It is represented by and, or, not.

```python
print((10>5)and(2<3))
print((2<1)or(10>11))
print(not(10>11)and(9<11))
```

```python
print((10<11)and(11>12))
print((2<3)or(10>11))
print((10>11)or not(10<1))
```

(e.)    Bitwise operators --------> These ar the operators that work at bit-level

```python
a=5
b=3

##### perform bitwise AND
print(a&b)

##### perform bitwise OR
print(a|b)

##### perform bitwise XOR
print(a^b)

##### perform bitwise NOT
print(~a)

##### perform bitwise left shift
print(a<<1)

##### perform bitwise right shift
print(a>>1)
```

(f.) Membership operator -------> It is used to check whether the particular element is a member of the given collection

```python
print(1 in [1,2,3])
print(5 in [1,2,3])

print(10 not in [1,2,3])
print(11 not in [11,12,13])
```

##### Analytical Business Scenario Questions

1.  Employee Payroll Processing

You are working as a Data Engineer in a company. You receive an employee dataset containing the following columns:


|Employee |	Experience	| Monthly Salary  |	Performance   |	  Status     |
|-------- | ------------|---------------  |---------------|--------------|
|Amit	  |        5	|        50000	  |      4.5	  |   Active     |
|Rahul	  |        2	|        35000	  |       3.0	  |   Active     |
|Priya	  |        6	|        65000	  |       4.8	  |   Active     |
|Sneha	  |        1	|        25000	  |       2.5	  |   Inactive   |
|Arjun	  |        4	|        45000	  |      4.0	  |   Active     |


Business requirements:
- Consider only active employees.
- Employees with at least 3 years of experience are eligible for a 10% salary increment.
- Employees with performance ratings of 4 or above receive an additional ₹5,000 bonus.
- Calculate the revised monthly salary.
- Identify employees whose revised monthly salary exceeds ₹50,000.
- Check whether the employee name exists in the employee list

```python
emp_details = [
    {"Employee":"Amit" , "Experience":5, "Monthly Salary" :50000, "Performance":4.5,"Status":"Active"},
    {"Employee":"Rahul", "Experience":2, "Monthly Salary":35000, "Performance":3.0, "Status":"Active"},
    {"Employee":"Priya", "Experience":6, "Monthly Salary":65000, "Performance":4.8, "Status":"Active"},
    {"Employee":"Sneha", "Experience":1, "Monthly Salary":25000, "Performance":2.5, "Status":"Inactive"},
    {"Employee":"Arjun", "Experience":4, "Monthly Salary":45000, "Performance":4.0, "Status":"Active"}
    ]

emp_details
```

```python
##### Consider only active employees
for x in emp_details:
    if(x['Status']=='Active'):
        print(x)
```

```python
##### Employees with at least 3 years of experience are eligible for a 10% salary increment
for x in emp_details:
    if(x['Experience'] >= 3):
        x['Monthly Salary'] += (0.10)*x['Monthly Salary']
        print(x)
```

```python
##### Employees with performance ratings of 4 or above receive an additional ₹5,000 bonus
for x in emp_details:
    if(x['Performance'] >= 4):
        x['Monthly Salary'] += 5000
        print(x)
```

```python
##### Calculate the revised monthly salary
for x in emp_details:
    if(x['Status'] == 'Active'):
        increment=0
        bonus=0
        if(x['Experience']>=3):
            increment = (0.10)*x['Monthly Salary']
        else:
            increment = 0
        if(x['Performance']>=4):
            bonus=5000
        else:
            bonus=0
        
        ##### Calculate the revised salary
        revised_salary = x['Monthly Salary'] + increment + bonus
        print(revised_salary)


        ##### update the revised salary to the monthly salary
        x['Monthly Salary'] = revised_salary
        
print(emp_details)
```

```python
##### Identify employees whose revised monthly salary exceeds ₹50,000
for x in emp_details:
    if(x['Monthly Salary']>50000):
        print(x)
```

```python
##### Check whether the employee name exists in the employee list
for x in emp_details:
    if(x['Employee'] == 'Amit'):
        print(f'{x['Employee']} exists in the employee list')
```
