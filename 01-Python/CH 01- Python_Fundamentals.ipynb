4. Write a Python program to calculate the total and average marks of five subjects

```python
m1 = 85
m2 = 87
m3 = 95
m4 = 98
m5 = 90

total = m1+m2+m3+m4+m5
print(f'Total marks is {total}')

avg = total/5
print(f'Average marks is {avg:.2f}')
```
