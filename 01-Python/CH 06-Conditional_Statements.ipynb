##### Conditional Statements

1.  Conditional statements are the statements that is used to make some decisions based on some conditions.

##### Types of Conditional Statements

1.  if statement

    This statement executes the set of statements if the condition is true else does nothing.

```python
##### Check if the number is positive
num=10
if(num>=10):
    print(f"{num} is positive")
```

2. if else statement

   This statement executes the set of statements if the condition is True else it executes another set of statements.

```python
##### Check if the number is positive or negative
num=10
if(num>0):
    print(f'{num} is positive')
else:
    print(f'{num} is negative')
```

3. elif statements

   This statements executes one set of statements if one condition is true else it check for another condition then executes another set of statements else executes another set of statements

```python
##### Check if the number is positive or negative or zero
num=10
if(num>0):
    print(f'{num} is positive')
elif(num<0):
    print(f'{num} is negative')
else:
    print(f'{num} is zero')
```

4. Nested if statements

    When one if statements is enclosed within another if statements then it is known as nested if statements.

```python
##### Check if the number is even positive or negative 
num=10
if(num>0):
    if(num%2==0):
        print(f"{num} is even positive")
    else:
        print(f"{num} is odd positive")
else:
    print(f"{num} is negative")
```

##### Analytical Scenario-Based Question

1.  Employee Payroll Processing System

You are working as a Data Engineer in a company. You receive an employee dataset containing the following columns:

|Employee |	Experience|	Monthly Salary |	Performance |	Status   |
|---------|-----------|----------------|----------------|------------|
|Amit	  |    5	  |     50000	   |      4	        |   Active   |
|Rahul	  |    2	  |     35000	   |      3	        |   Active   |
|Priya	  |    8	  |     75000	   |      5	        |   Active   | 
|Sneha	  |    1	  |     25000	   |      2	        |   Inactive |
|Arjun	  |    6	  |     60000	   |      3	        |   Active   |

Business requirements:
- Process only active employees.
- Employees with at least 5 years of experience are eligible for a bonus.
- Employees with performance rating 4 or 5 receive a 20% bonus.
- Employees with performance rating 3 receive a 10% bonus.
- Employees with performance rating below 3 receive no bonus.
- Calculate the final salary after adding the bonus.

```python
emp_details = [
    {"Employee":"Amit" , "Experience":5, "Monthly Salary" :50000, "Performance":4,"Status":"Active"},
    {"Employee":"Rahul", "Experience":2, "Monthly Salary":35000, "Performance":3, "Status":"Active"},
    {"Employee":"Priya", "Experience":8, "Monthly Salary":75000, "Performance":5, "Status":"Active"},
    {"Employee":"Sneha", "Experience":1, "Monthly Salary":25000, "Performance":2, "Status":"Inactive"},
    {"Employee":"Arjun", "Experience":6, "Monthly Salary":60000, "Performance":3, "Status":"Active"}
    ]

emp_details
```

```python
bonus=0
for x in emp_details:
    if(x['Status']=='Active'):
        if(x['Experience']>=5):
            if(x['Performance']==4 or x['Performance']==5):
                bonus = (0.20)*x['Monthly Salary']
            elif(x['Performance']==3):
                bonus = (0.10)*x['Monthly Salary']
            elif(x['Performance']<3):
                bonus=0
        else:
            bonus=0
        final_salary = x['Monthly Salary']+bonus

        print('Employee name :', x['Employee'])
        print('Experience    :', x['Experience'])
        print('Monthly Salary:', x['Monthly Salary'])
        print('Performance   :', x['Performance'])
        print('Status        :' ,x['Status'])
        print('Final Salary  :' ,final_salary)
        print('----------------------------------------')
```
