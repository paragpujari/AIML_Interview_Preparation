##### Loops

1.  A Loop helps in executing a block of code unless and until a specific condition is satisfied.

##### Types of Loop

(a.)    for loop        -----> Iterates over a sequence (list, tuple, string, range, etc.)
  
(b.)    while loop      -----> Executes repeatedly as long as a condition is True

##### Solved Examples

1.  Print numbers from 1 to 10 using for loop

```python
for i in range(1,11):
    print(i)
```

2.  Calculate the sum of numbers from 1 to 10

```python
s=0
for i in range(1,11):
    s+=i
print(f'Sum of all the numbers from 1 till 10 is {s}')
```

3. Print even numbers from 1 to 20

```python
for i in range(1,21):
    if(i%2==0):
        print(i)
```

4. Calculate the factorial of a number using while loop

```python
f=1
i=1
while(i<=5):
    f*=i
    i+=1
print(f'Factorial of the number is {f}')
```

5. Find the sum of digits of a number

```python
s=0
num=int(input("Enter the number"))
x=num
while(x!=0):
    d=x%10
    s+=d
    x=x//10
print(f'Sum of all the digits of the number is {s}')
```

6. Print all the prime numbers within a given range

```python
a=10
b=100
for i in range(a,b+1):
    flag=True
    if(i>1):
        for j in range(2,i):
            if(i%j==0):
                flag=False
                break

        if(flag==True):
            print(i)
```

8. Write a Python program to check whether a given number is a palindrome using a loop

```python
num=121
x=num
rev=0
while(x!=0):
    d=x%10
    rev=(rev*10)+d
    x=x//10
if(rev==num):
    print(f"{num} is palindrome")
else:
    print(f"{num} is not palindrome")
```

8. Write a Python program to check whether a given number is an armstrong or not

```python
num=153
x=num
s=0
while(x!=0):
    d=x%10
    s=s+(d*d*d)
    x=x//10
if(s==num):
    print(f"{num} is an armstrong")
else:
    print(f"{num} is not an armstrong")
```

1.  A company maintains employee information in the following format:

Business Requirements:

Write a Python program using loops to:
- Iterate through all employee records.
- Identify employees whose status is Active.
- Calculate a 10% bonus for active employees.
- Calculate the final salary (Salary + Bonus).
- Display employee name, original salary, bonus and final salary.
- Calculate the total bonus paid by the company.
- Count the total number of active employees.

```python
##### Iterate through all employee records
employees = [    {"Name": "Amit", "Salary": 50000, "Status": "Active"},    {"Name": "Rahul", "Salary": 60000, "Status": "Inactive"},    {"Name": "Priya", "Salary": 70000, "Status": "Active"},    {"Name": "Sneha", "Salary": 45000, "Status": "Active"},    {"Name": "Ravi", "Salary": 55000, "Status": "Inactive"}]

for x in employees:
    print(x)
```

```python
##### Identify employees whose status is Active
for x in employees:
    if(x['Status']=='Active'):
        print(x)
```

```python
##### Calculate a 10% bonus for active employees
for x in employees:
    if(x['Status']=='Active'):
        bonus = (0.10)*x['Salary']
        ##### Calculate the final salary (Salary + Bonus)
        final_salary = x['Salary'] + bonus
        ##### Display employee name, original salary, bonus and final salary
        print(f'Employee name is {x['Name']}')
        print(f'Original salary is {x['Salary']}')
        print(f'Bonus is {bonus}')
        print(f'Final salary is {final_salary}')
        print('-----------------------------------')
```

```python
##### Calculate the total bonus paid by the company
##### Count the total number of active employees
tb=0
c=0
for x in employees:
    if(x['Status']=='Active'):
        ##### Calculate the bonus on active employees
        bonus = (0.10)*x['Salary']
        ##### Calculate the final salary
        final_salary = x['Salary'] + bonus
        ##### Calculate the total bonus
        tb = tb + bonus
        ##### Calculate the total number of active employees
        c  = c  +  1
        ##### Display employee name, original salary, bonus and final salary
        print(f'Employee name is {x['Name']}')
        print(f'Original salary is {x['Salary']}')
        print(f'Bonus is {bonus}')
        print(f'Final salary is {final_salary}')
        print('--------------------------------')

print(f'Total bonus paid by the company is {tb}')
print(f'Total number of active employees is {c}')
```
