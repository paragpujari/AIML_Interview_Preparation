##### Type Conversions

1.  Type conversion refers to the technique of converting the value from one data type to another data type.

##### Types of Type Conversion

(a.)    Implicit Type Conversion    ---------->

It is the technique of converting the value from one data type to another automatically.

```python
x = 10
y = 5.7

x = x+y
print(x)
```

##### OBSERVATIONS:

1.  Here x was initially 10 (int) then after performing addition of 5.7 , x became 15.7(float).

2.  So by performing normal addition, the value of x got changed from int to float.

(b.)    Explicit Type Conversion    ---------->

It is the technique of manually converting the value from one type to another.

```python
x = '10'

y = 20

res = int(x) + y

print(res)
```

##### OBSERVATIONS:

1.  Initially, the value of x was '10' in string form. Then we have y as 20 in integer form.

2.  On performing addition of string with int, it will show an error.

3.  To avoid the error, string form of int is converted into int using int() function.

##### Properties of Type Conversions

1.  On converting a value from int(lower) -------> float(higher) data type, there is no loss of precision. No loss of data takes place. It is safe to use.

2.  On converting the value from float(higher) ----> int(lower) data type, there is loss of precision. Loss of data takes place. It is not safe to use.

##### Common Type Conversion functions

|   Function   |	    Conversion                                 |
| -------------|---------------------------------------------------|
|   int()	   |Converts into integer                              |
|   float()	   |Converts into float                                |
|   str()	   |Converts into string                               |
|   bool()	   |Converts into Boolean                              |
|   list()	   |Converts into list                                 |
|   tuple()	   |Converts into tuple                                |
|   set()	   |Converts into set                                  |
|   dict()	   |Converts suitable key-value pairs into dictionary  |

##### Solved Examples

1.  Integer to Float

```python
age=31
print(f'Original Age is {age}')
print(f'Original data type of age is {type(age)}')
print("-------------------------------------------------------------------")
##### convert the age from int to float
ans = float(age)
print(f'Converted Age is {ans}')
print(f'Converted data type of age is {type(ans)}')
```

2.  Float to Integer

```python
salary = 53874.35
print(f'Original salary is {salary}')
print(f'Original data type of salary is {type(salary)}')

print("-------------------------------------------------------------------")
##### Convert the salary from float to int
ans = int(salary)
print(f'Converted Salary is {ans}')
print(f'Converted data type of salary is {type(ans)}')
```

3.  Integer to String

```python
emp_id = 10
print(f'Original employee id is {emp_id}')
print(f'Original data type of the employee id is {type(emp_id)}')
print("-------------------------------------------------------------------")

##### Convert the employee id from integer to string
ans = str(emp_id)
print(f'Converted employee id is {ans}')
print(f'Converted data type of the employee id is {type(ans)}')
```

4. String to Integer

```python
price = "2500"
print(f'Original product price is {price}')
print(f'Original data type of product price is {type(price)}')
print("-------------------------------------------------------------------")

ans = int(price)
print(f'Converted product price is {ans}')
print(f'Converted data type of product price is {type(ans)}')
```

5.  List to Tuple and Set

```python
product_list = ["Laptop","Smartphone","Tablet"]

##### convert to tuple
product_tuple = tuple(product_list)
print(product_tuple)
print(type(product_tuple))
print("-------------------------------------------------------------------")
##### convert to set
product_set = set(product_list)
print(product_set)
print(type(product_set))
```
