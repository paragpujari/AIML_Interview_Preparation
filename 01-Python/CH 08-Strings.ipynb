22. A company wants to identify duplicate characters in customer usernames.

Write a Python program to identify all characters that appear more than once in a given string

```python
x="programming"
##### convert the text to lower case
x=x.lower()
#print(x)

freq={}
for i in x:
    if(i in freq):
        freq[i] = freq[i]+1
    else:
        freq[i] = 1

for x,y in freq.items():
    if(y>1):
        print(x,end=',')
```

23.  A company requires passwords to contain:
- At least one uppercase character.
- At least one lowercase character.
- At least one digit.
- Minimum length of 8 characters.

Write a Python program to validate whether a given password satisfies all conditions

```python
password = "Python123"
upper=False
lower=False
digit=False
for x in password:
    if(x.isupper()):
        upper=True
    elif(x.islower()):
        lower=True
    elif(x.isdigit()):
        digit=True

##### To check if the password is valid or invalid
if((upper==True)and(lower==True)and(digit==True)and(len(password)>=8)):
    print("Valid password")
else:
    print("Invalid password")
```

30. Write a Python program to check whether the given text is a palindrome after:
- Converting all characters to lowercase.
- Removing spaces and special characters.
- Comparing the cleaned string with its reverse.

```python
text = "A man, a plan, a canal: Panama"
##### convert all the characters to lowercase
text=text.lower()
print(text)
##### Removing spaces and special characters
r=''
for i in text:
    if(i==' ' or i==',' or i==':'):
        continue
    #print(i,end='')
    r=r+i
print(r)

##### Get the reverse of the string
print(r)
x=r[::-1]
print(x)

if(r==x):
    print(f'{r} is a palindrome')
else:
    print(f'{r} is not a palindrome')
```

37. Compress consecutive characters

```python
x="aaabbc"
x=x.lower()
c=1
res=''
for i in range(1,len(x)):
    if(x[i]==x[i-1]):
        c=c+1
    else:
        res=res+x[i-1]+str(c)
        c=1
res=res+x[i-1]+str(c)
print(res)
```
