##### Properties of Variable

1.  A Variable in Python is dynamically typed. The data type of the variable is determined automatically once the value is assigned to the variable.The value of the variable gets changed at different runtime execution of the program.

```python
x = 10
print(x)

x =70.76
print(x)

x = True
print(x)

x = (4+5j)
print(x)

x = 'Shreyash'
print(x)
```

##### OBSERVATIONS:

1.  Here different values for x gets assigned to x at different runtime execution of the programs.

2.  The data type of x gets changed at different runtime execution as different values gets assigned to x at different instances.

##### Rules for defining the  Variable

1.  A Variable must have letters, numbers or underscore

```python
name_='Parag'   ##### A variable must have an underscore
num1 = 10       ##### A variable must have a  number
x = 10          ##### A variable must have a  letter

print(name_)
print(num1)
print(x)
```

2. A variable must begin with letter or an underscore

```python
x=10                ##### A Variable begins with a letter
print(x)

_x = 10             ##### A Varibale begins with an underscore
print(_x)
```

3.  A Variable must have underscore if needed

```python
first_name = 'Parag'
print(first_name)

last_name = 'Pujari'
print(last_name)
```

4. A Variable must not begin with digit

```python
# 1num=10    ##### A Variable should not begin with a digit
```
