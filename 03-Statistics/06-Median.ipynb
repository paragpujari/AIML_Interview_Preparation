##### Median

Median is the middle value of a dataset when all observations are arranged in ascending or descending order.





##### Solved Examples

1.  Find the median

```python
import numpy as np

data = [10, 20, 30, 40, 50]

##### Find out the median from the above series
ans = np.median(data)
print(f'Median of the above series is {ans}')
```

##### Analytical Scenario Based Questions

1. A company wants to analyze employee salaries to understand the typical salary of its workforce.

The HR department provides the following salary dataset

Questions:
- Calculate the median employee salary.
- Calculate the mean employee salary.
- Why is the mean significantly higher than the median?
- Which measure (Mean or Median) better represents the typical employee salary?
- How does the unusually high salary of ₹2,00,000 affect the mean and median differently?

```python
salary = [
    25000,
    28000,
    30000,
    32000,
    35000,
    38000,
    40000,
    45000,
    50000,
    200000
]

##### Calculate the median employee salary

ans_median = np.median(salary)

print(f'Median of the employee salary dataset is {ans_median}')

##### Calculate the mean employee salary

ans_mean = np.mean(salary)

print(f'Mean of the employee salary dataset is {ans_mean}')
```

##### Why is the mean significantly higher than the median

Ans Mean is significantly higher than the median because:

    (a.)    In mean, every value is taken into consideration. If there is an outlier, then the value is highly affected by the outliers.

    (b.)    In median, it first sorts the dataset, then considers the middle element. The median element remains unaffected by the outlier.

So for this reason, the median is significantly higher than the median.

##### Which measure (Mean or Median) better represents the typical employee salary

Ans Median better represents the above employee salary dataset as it remains unaffected by the outliers.

    It provides the accurate value even if outlier is present in the data.

##### How does the unusually high salary of ₹2,00,000 affect the mean and median differently

(a.) Mean


```python
salary = [
    25000,
    28000,
    30000,
    32000,
    35000,
    38000,
    40000,
    45000,
    50000,
    200000
]

ans = np.mean(salary)
print(ans)



##### Without outlier
salary = [
    25000,
    28000,
    30000,
    32000,
    35000,
    38000,
    40000,
    45000,
    50000
]

ans = np.mean(salary)
print(ans)
```

##### OBSERVATIONS:

-  On considering the outlier in the data series, the mean value is 52300.

- On removing the outlier from the data series, the mean value becomes 35888 which is nearly equal to the median value(36500)

(b.) Median

```python
salary = [
    25000,
    28000,
    30000,
    32000,
    35000,
    38000,
    40000,
    45000,
    50000,
    200000
]

ans = np.median(salary)
print(ans)
```

##### OBSERVATIONS:

- On considering the outlier in the above data series, the median value is 36500 which is unaffected by the extremely large value of 200000.
