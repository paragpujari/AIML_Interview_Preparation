##### Mode

Mode is a technique used in the measures of central tendency that is used to find out the most frequently occuring element from the given data.

##### Types of Modes:

-Unimodal: Dataset has one mode.

-Bimodal: Dataset has two modes.

-Multimodal: Dataset has more than two modes.

-No mode: No value repeats (all values occur equally often)

##### Five Examples

1.  Find the mode of a list

```python
data = [10, 20, 30, 20, 40, 20, 50]

##### Find the mode of the above data
from statistics import mode

ans = mode(data)

print(f'Mode of the above data series is {ans}')
```

2. Find the mode of the given list using numpy

```python
import numpy as np

datas = [10, 20, 30, 20, 40, 20, 50]

##### Calculate the mode of the above data set
data,frequencies = np.unique(datas,return_counts=True)

print(data)
print(frequencies)

##### Get the index value for the maximum occuring element in the frequencies
index=np.argmax(frequencies)
print(index)

##### Get the value at the index position of the maximum occuring element
ans=data[index]
print(f'Mode of the above data series is {ans}')
```

3. Find the mode of student marks

```python
marks = [70, 80, 90, 80, 60, 80, 90, 70]

from statistics import mode

ans = mode(marks)

print(f'Mode of the above student marks is {ans}')
```

4. Find multiple modes

```python
data = [10, 20, 20, 30, 30, 40, 50]

##### Find out the multiple modes from the above dataset

from statistics import multimode

ans = multimode(data)

print(f'Multiple number of modes from the above dataset is {ans}')
```

5. Find the most frequent element in a list

```python
data = [2, 4, 3, 5, 2, 8, 2, 4, 4, 4]

from collections import Counter

##### To find out theb frequency of each element present in the list
frequency = Counter(data)

###### Arrange all the frequencies of the elements in descending order
ans = frequency.most_common(1)

print(f'Most frequent element in a list is {ans}')
```

4. Write a Python program to identify whether a dataset is unimodal, bimodal, or multimodal

```python
from collections import Counter

data = [10, 20, 20, 30, 30, 40, 50]

frequencies = Counter(data)

print(frequencies)

##### To find out the maximum frequency
max_frequency = max(frequencies.values())
print(max_frequency)

frequencies = dict(frequencies)
print(frequencies)
##### To classify the type of mode
lists=[]
for x,y in frequencies.items():
    if(y==max_frequency):
        print(x)
        lists.append(x)
        print(lists)

if(len(lists)==1):
    print(f"Unimodal")
elif(len(lists)==2):
    print(f"Bimodal")
elif(len(lists)>2):
    print(f"Multi modal")
else:
    print("No modes")
```

10. Write a Python program to find the most frequently occurring word in a given sentence

```python
from collections import Counter

sentence = "python is easy and python is powerful and python is popular"

##### Break the sentence into words
words = sentence.split(" ")
print(words)

##### Get the frequencies of all elements in the text

frequencies = Counter(words)

print(frequencies)

##### Most frequently occuring word
ans = frequencies.most_common(1)
print(f'Most frequently occuring word in the sentence is {ans}')
```

(c.) What does the mode tell HR about the organization's salary distribution?

(d.)  If the organization increases the salary of one employee earning ₹40,000 to ₹1,00,000, will the mode change?

```python
from statistics import mode 

salaries = [
    30000, 100000, 50000, 40000,
    60000, 40000, 70000, 50000,
    40000, 80000, 50000, 90000
]

ans = mode(salaries)

print(ans)


from collections import Counter

f2 = Counter(salaries)

print(f2)
```

##### OBSERVATIONS:

- On changing the salary of one employee from 40000 to 10000, we came to know that

    - Initially 40000 appeared 4 times, but 50000 appeared 3 times. So it was a unimodel. 

    - After the salary of one employee got changed from 40000 to 10000, both 40000 and 50000 appeared for 3 times. So it is a bimodel.
