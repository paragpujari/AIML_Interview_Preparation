##### Mean

1.  Mean -----> 

It can be defined as one of the measures of central tendency that is used to find out the average of all the observations.

It can be also defined as the sum of all the observations divided by total number of observations.

##### Formula:



```python
##### Find out the mean of below five elements in the list
lists = [10,20,30,40,50]

##### Calculate the mean of all the elements in the list
mean = sum(lists)/len(lists)

print(f'Mean of all the elements in the list is {mean}')
```

##### Types of Mean

|Type	         |       Meaning                         |
|----------------|---------------------------------------|
|Arithmetic Mean | 	Sum of values / Number of values     |
|Weighted Mean	 |  Average considering weights          |
|Geometric Mean	 |  Based on product of values           |
|Harmonic Mean	 |  Reciprocal-based average             |

1.  Find the average of 10, 20, 30, 40, 50

3. Calculate mean using a for loop

```python
numbers = [15, 25, 35, 45, 55]

total=0
for i in numbers:
    total += i

l = len(numbers)

##### Calculate the mean of all the elements in the list
mean = total/l
print(f"Mean of all the elements in the list is {mean:.2f}")
```

4. Calculate the average salary of active employees

```python
employees = [
    {"Name": "Amit", "Salary": 50000, "Status": "Active"},
    {"Name": "Rahul", "Salary": 40000, "Status": "Inactive"},
    {"Name": "Priya", "Salary": 60000, "Status": "Active"},
    {"Name": "Neha", "Salary": 70000, "Status": "Active"}
]

##### Calculate the average salaries of all the active employees
s=0
c=0

for x in employees:
    if(x["Status"]=="Active"):
        s=s+x["Salary"]
        c=c+1

##### Calculate the average salary
mean = s/c
print(f"Average salaries of all active employees is {mean:.2f}")
```

5. Calculate a student's weighted average based on marks and subject credits

|Subject    |	Marks	|Credits |
|-----------|-----------|--------|
|Python	    |   80	    |   4    |
|SQL	    |   90	    |   3    |
|Statistics |	70	    |   2    |

```python
students = [
    {"Subject": "Python"    , "Marks":  80,"Credits": 4},
    {"Subject": "SQL"       , "Marks": 90, "Credits": 3},
    {"Subject": "Statistics", "Marks": 70, "Credits": 2}
]

print(students)

weighted_sum=0
total_credits=0
###### Calculate the weighted mean
for x in students:
    x["Total"] = x["Marks"] * x["Credits"]
    weighted_sum += x["Total"]
    total_credits += x["Credits"]

mean = weighted_sum/total_credits
print(f"Weighted mean ofn Student's marks and credits is {mean:.2f}")
    
```

##### Analytical Scenario Based Questions

1.  Scenario: Employee Salary Analysis
A company has the following employee records:

|Employee |	Department	|   Salary   |
|---------|-------------|------------|
|Amit	  |     IT	    |   60,000   |
|Rahul    |	    HR	    |   40,000   |
|Priya	  |     IT	    |   80,000   |
|Neha	  |     Finance	|   50,000   |
|Kiran	  |     IT	    |   70,000   |
|Sneha	  |     HR	    |   30,000   |

Business requirements:
- Calculate the overall average employee salary.
- Calculate the average salary of the IT department.
- Calculate the average salary of the HR department.
- Identify employees earning above the overall average salary.
- Identify which department has the highest average salary.

```python
employees = [
    {"Employee": "Amit",  "Department": "IT",      "Salary": 60000},
    {"Employee": "Rahul", "Department": "HR",      "Salary": 40000},
    {"Employee": "Priya", "Department": "IT",      "Salary": 80000},
    {"Employee": "Neha",  "Department": "Finance", "Salary": 50000},
    {"Employee": "Kiran", "Department": "IT",      "Salary": 70000},
    {"Employee": "Sneha", "Department": "HR",      "Salary": 30000},
]

##### Calculate the overall average employee salary
s=0
c=0
for x in employees:
    s+=x['Salary']
    c+=1
mean = s/c
print(f'Average of all the employee salary is {mean:.2f}')
```

```python
##### Calculate the average salary of the IT department
s=0
c=0
for x in employees:
    if(x["Department"]=="IT"):
        s+=x['Salary']
        c+=1
mean=s/c
print(f'Average salary of the IT department is {mean:.2f}')
```

```python
##### Calculate the average salary of the HR department
s=0
c=0
for x in employees:
    if(x["Department"]=="HR"):
        s+=x['Salary']
        c+=1
mean=s/c
print(f'Average salary of the IT department is {mean:.2f}')
```

```python
##### Identify employees earning above the overall average salary
s=0
c=0
for x in employees:
    s+=x['Salary']
    c+=1

mean=s/c
print(f'Mean of all the salaries is {mean:.2f}')

for x in employees:
    if(x['Salary']>mean):
        print(x)
```

```python
##### Identify which department has the highest average salary
import pandas as pd

##### convert the list into dataframe
df = pd.DataFrame(employees)
print(df)

##### Calculate the mean salary of each department
x=df.groupby('Department')['Salary'].mean()
print(x)

##### order the salaries in descending order
ans = x.sort_values(ascending=False)
print(ans)
```

##### OBSERVATIONS:

(a.)    IT department has the highest average salary.
