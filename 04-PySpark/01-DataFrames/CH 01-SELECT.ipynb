**Select()**

1.  select() statement helps in fetching certain columns from the Pyspark Dataframe.

**Syntax of Select()**

df.select("column1","column2")

**Example**

1.  Write a Pyspark Dataframe for the employee dataset

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,sum,avg
##### Create the session for Pyspark
session = SparkSession.builder.appName('select').getOrCreate()
##### get the version of spark
print(session.version)
```

```python
employees = [
    (1, "Amit", "IT", 60000, "Hyderabad"),
    (2, "Rahul", "HR", 45000, "Mumbai"),
    (3, "Priya", "IT", 75000, "Bangalore"),
    (4, "Sneha", "Finance", 55000, "Hyderabad"),
    (5, "Arjun", "Finance", 80000, "Delhi")
]

columns = ["emp_id","emp_name","department","salary","city"]

##### Create the dataframe from the above data
df = session.createDataFrame(employees,columns)

##### display the dataframe
df.show()
```

**Solved Examples**

1. Display only the employee names

```python
df.select('emp_name').show()
```

2. Display employee name, department and salary

```python
df.select('emp_name','department','salary').show()
```

3. Display all employee columns using select()

```python
df.select('*').show()
```

4. Display employee name and salary, renaming salary as monthly_salary

```python
df.select(col('emp_name'),col("salary").alias('monthly_salary')).show()
```

5. Display employee name, monthly salary and annual salary

```python
df.select(col('emp_name'),col('salary').alias('monthly_salary'),(col('salary')*12).alias('annual_salary')).show()
```

6. Display emp_id and name, but rename emp_id as employee_id

7. Display employee name, department and salary with the following aliases:
- name → employee_name
- department → dept
- salary → monthly_salary

```python
df.select(col('emp_name').alias('employee_name'),col('department').alias('dept'),col('salary').alias('monthly_salary')).show()
```

8. Display employee name and annual salary. Rename the calculated column as annual_salary

```python
df.select(col('emp_name').alias('employee_name'), (col('salary')*12).alias('annual_salary')).show()
```

9. Display employee name, current salary and salary after a 10% increment. Rename the calculated column as revised_salary.

```python
df.select(col('emp_name'),col('salary'),(col('salary')*1.10).alias('revised_salary')).show()
```

10. Display:
- Employee name
- Monthly salary
- Annual salary
- Monthly salary after a 15% increment
- Annual salary after a 15% increment

```python
df.select(col('emp_name'),col('salary').alias('Monthly_salary'),(col('salary')*12).alias('Annual_salary'),(col('salary')*1.15).alias('revised_monthly_salary'),(col('salary')*12*1.15).alias('revised_annual_salary')).show()
```

**Analytical Scenario Based Question**

1.  You are working as a Data Engineer in an HR analytics project.

The HR department wants a specific employee compensation dataset containing the following information:
- Employee ID
- Employee name
- Department
- Current monthly salary
- Annual salary
- Salary after a 12% increment
- Annual salary after a 12% increment

```python
df.select(col('emp_id').alias('Employee_id'),
          col('emp_name').alias('Employee_name'),
          col('department').alias('Department'),
          col('salary').alias('Current_monthly_salary'),
          (col('salary')*12).alias('Annual_salary'),
          (col('salary')*1.12).alias('salary_increment'),
          (col('salary')*12*1.12).alias('Annual_salary_increment')).show()
```
