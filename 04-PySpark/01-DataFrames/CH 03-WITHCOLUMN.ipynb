**WithColumn**

It is used to create a new column or modify an existing column in the PySpark Dataframe.

**Symtax:**

df.withColumn("column_name", expression)

##### Creation of PySpark Session

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

##### Creation of PySpark Session
session = SparkSession.builder.appName("withcolumn").getOrCreate()

print(session.version)
```

**Creation of Spark Dataframe**

```python
data = [(101, "Amit", 50000),
        (102, "Rahul", 60000)]

columns = ["id", "name","salary"]

df = session.createDataFrame(data,columns)

df.show()
```

**Solved Examples**

1.  Add a new column Bonus with a fixed value of 5000

```python
from pyspark.sql.functions import lit
##### Add a new column bonus and add 5000 value in it
df.withColumn('bonus',lit(5000)).show()
```

2. Create a new column Annual_Salary by multiplying monthly salary by 12

```python
df.withColumn('Annual_Salary',col('salary')*12).show()
```

3. Increase every employee's salary by 10%

```python
df.withColumn('Revised_Salary',col('salary')*1.10).show()
```

4. Create a column Salary_Status:
- Salary greater than or equal to 60000 → High
- Otherwise → Low

```python
df.withColumn('Salary_Status', when(col('salary')>=60000, "High").otherwise("Low")).show()
```

5. Create a new column Upper_Name containing uppercase employee names

```python
df.withColumn('Upper_Name',upper(col('name'))).show()
```

**Coding Interview Questions**

Use the following DataFrame:

```python
data = [
    (101, "Amit", "IT", 50000, 25),
    (102, "Rahul", "HR", 60000, 30),
    (103, "Priya", "Finance", 70000, 28),
    (104, "Sneha", "IT", 45000, 24),
    (105, "Vijay", "Sales", 80000, 35)
]

df = spark.createDataFrame(
    data,
    ["emp_id", "name", "department", "salary", "age"]
)

df.show()
```

1. Create a new column Annual_Salary by multiplying monthly salary by 12

```python
df.withColumn('Annual_Salary',col('salary')*12).show()
```

2. Create a new column Bonus by calculating 15% of the employee's salary

```python
df.withColumn('Bonus',col('salary')*0.15).show()
```

3. Create a new column Salary_Status where salary greater than 60000 is classified as High, otherwise Low

```python
df.withColumn('Salary_Status',when(col('salary')>60000,"High").otherwise("Low")).show()
```

4. Create a new column Experience_Category based on age:
- Age greater than 30 → Senior
- Otherwise → Junior

```python
df.withColumn('Experience_Category',when(col('age')>30,"Senior").otherwise("Junior")).show()
```

5. Create a new column Employee_Name containing employee names in uppercase

```python
df.withColumn('Employee_Name',upper(col('name'))).show()
```

6. Increase every employee's salary by 20% by modifying the existing salary column

```python
df.withColumn('Modified_Salary',col('salary')*1.2).show()
```

7. Create a new column Department_Label by adding the prefix Dept_ to each department name

```python
df.withColumn('Department_Label',concat(lit('Dept_'),col('department'))).show()
```

8. Create a new column Age_After_5_Years by adding 5 to the existing age

```python
df.withColumn('Age_After_5_Years',lit(5)+col('age')).show()
```

9. Create a new column Tax by calculating 10% of the employee's salary

```python
df.withColumn('Tax',col('salary')*0.1).show()
```

10. Create a new column Employee_Category:
- Salary greater than or equal to 70000 → Premium
- Salary between 50000 and 69999 → Standard
- Salary below 50000 → Basic

```python
df.withColumn(
    'Employee_Category'         ,
    when(col('salary')>=70000,'Premium')
    .when((col('salary')>=50000)&(col('salary')<=69999), 'Standard')
    .otherwise('Basic')
    ).show()
```

**Analytical scenario-based interview question**

1. A company maintains an employee DataFrame with the following columns:
- emp_id
- emp_name
- department
- salary
- age
The HR department wants to enhance its employee DataFrame using only PySpark's withColumn() statement.

Requirements:
1. Create a new column Annual_Salary by multiplying monthly salary by 12.
2. Create a new column Bonus equal to 20% of monthly salary.
3. Modify the existing salary column by increasing it by 10%.
4. Create a new column Salary_Grade:
   - Salary greater than or equal to 80000 → A
   - Salary between 50000 and 79999 → B
   - Salary below 50000 → C
5. Create a new column Employee_Name containing uppercase employee names

```python
df.withColumn('Annual_Salary',col('salary')*12).withColumn('bonus',col('salary')*0.20).withColumn('modified_salary',1.10*col('salary')).withColumn('Salary_Grade',when(col('salary')>=80000, 'A')
                           .when((col('salary')>=50000)&(col('salary')<=79999),'B')
                           .otherwise('C')
                           ).show()
```
