**withColumnRenamed()**

    - It is used to rename an existing column name with a new column name in the PySpark Dataframe.

**Syntax**

    - df.withColumnRenamed("old_col_name","new_col_name")

**Creation of Spark Session**


```python
from pyspark.sql import SparkSession

##### Create the session for spark]
session = SparkSession.builder.appName("columnrenamed").getOrCreate()

print(session.version)
```

**Create the data**

```python
data = [
    (101, "Amit", 50000),
    (102, "Rahul", 60000),
    (103, "Priya", 70000)
]

columns = ["emp_id","emp_name","salary"]

##### Create the dataframe from the above data
df = session.createDataFrame(data,columns)

df.show()
```

**Solved Examples**

1.  Rename a single column emp_name to employee_name

```python
df.withColumnRenamed('emp_name','employee_name').show()
```

2. Rename multiple columns

```python
df.withColumnRenamed('emp_id','employee_id')\
  .withColumnRenamed('emp_name','employe_name')\
  .withColumnRenamed('salary','employee_salary').show()
```

3. Rename a column containing 'emp_name' to uppercase letters

```python
df.withColumnRenamed('emp_name','EMP_NAME').show()
```

4. Rename a column to column containing spaces

```python
df.withColumnRenamed('emp_name','emp name').show()
```

5. Rename a column and verify original DataFrame

```python
df1 = df.withColumnRenamed('salary','monthly_salary')

print("Original Dataframe")
df.show()

print("Modified Dataframe")
df1.show()
```
