**filter()**

It is a statement used in Pyspark that is used to fetch the records from the database table based on some conditions.

**Creation of Spark Session**

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,sum,avg

##### Creation of Spark Session
session = SparkSession.builder.appName('filter').getOrCreate()
print(session.version)
```

**Creation of DataFrame**

```python
data = [
    (1, "Amit", "IT", 60000, "Hyderabad", "Active"),
    (2, "Rahul", "HR", 40000, "Mumbai", "Inactive"),
    (3, "Priya", "IT", 75000, "Delhi", "Active"),
    (4, "Sneha", "Finance", 50000, "Hyderabad", "Active"),
    (5, "Arjun", "HR", 35000, "Delhi", "Inactive"),
    (6, "Neha", "Finance", 80000, "Mumbai", "Active"),
    (7, "Kiran", "IT", 45000, "Hyderabad", "Inactive"),
    (8, "Ravi", "Sales", 55000, "Delhi", "Active"),
    (9, "Pooja", "Sales", 65000, "Mumbai", "Active"),
    (10, "Vijay", "IT", 90000, "Hyderabad", "Active")
]

columns = ["emp_id", "emp_name", "department","salary", "city", "status"]

###### Creation of dataframe from the above dataset
df = session.createDataFrame(data,columns)

##### Display the dataframe
df.show()
```

**Solved Examples**

1.  Filter employees with salary greater than 50,000

```python
##### Filter all the employees whose salary > 50000
df.filter(df.salary > 50000).show()
```

2. Filter employees belonging to IT department

```python
df.filter(df.department == "IT").show()
```

3. Filter employees from Hyderabad

```python
df.filter(df.city=="Hyderabad").show()
```

4. Filter active employees earning more than 50,000

```python
df.filter((df.status=="Active")&(df.salary>50000)).show()
```

5. Filter employees whose salary is between 40,000 and 70,000

```python
df.filter((df.salary >= 40000)&(df.salary <= 70000)).show()
```

**Coding Interview Questions**

1.  Write a PySpark filter condition to display employees whose salary is greater than 60,000

```python
df.filter(df.salary > 60000).show()
```

2. Write a PySpark query to display employees working in the Finance department

```python
df.filter(df.department=="Finance").show()
```

3. Display employees who belong to the IT department and have a salary greater than 50,000

```python
df.filter((df.department=="IT")&(df.salary > 50000)).show()
```

4. Display employees who belong to either the HR or Finance department

```python
df.filter((df.department=="HR")|(df.department=="Finance")).show()
```

5. Display all employees whose employment status is not Active

```python
df.filter(col('status')!='Active').show()
```

6. Display employees who belong to Hyderabad, Mumbai, or Delhi

```python
df.filter(col('city').isin("Hyderabad","Mumbai","Delhi")).show()
```

7. Display employees who do not belong to the IT or HR departments

```python
df.filter(~col('department').isin("IT","HR")).show()
```

```python
df.filter((col('department')!="IT")&(col("department")!="HR")).show()
```

8. Display employees whose salary is between 45,000 and 80,000, inclusive

```python
df.filter((col("salary")>=45000)&(col("salary")<=80000)).show()
```

9. Display employees whose names start with the letter 'A'

```python
df.filter(col("emp_name").like("A%")).show()
```

```python
df.filter(col('emp_name').startswith('A')).show()
```

10. Display employees who:
- Belong to the IT or Finance department.
- Have a salary greater than 50,000.
- Have Active status.

```python
df.filter(((df.department=="IT")|(df.department=="Finance"))&(df.salary>50000)&(df.status=="Active")).show()
```
